## 2. Loading Sparsity Comparison (Su_2020, SEA_AD, Yost_2019)

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

FIGDIR = "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Figures/Loading Matrixs Sparsity"
os.makedirs(FIGDIR, exist_ok=True)


def hoyer_sparsity(v):
    """Compute Hoyer sparsity (entropy-like sparsity measure)."""
    v = np.abs(v)
    return (np.sqrt(len(v)) - np.linalg.norm(v, 1) / np.linalg.norm(v, 2)) / (np.sqrt(len(v)) - 1 + 1e-12)


def compute_gene_entropy(L):
    """Compute sparsity for each gene (row)."""
    return np.array([hoyer_sparsity(L[i, :]) for i in range(L.shape[0])])


def plot_loading_sparsity(load_paths, method_labels, title, save_name):
    """
    Compute Hoyer sparsity and draw publication-quality violin + transparent box plots.
    Saves to FIGDIR as a PDF.
    """

    sns.set_theme(style="whitegrid")
    assert len(load_paths) == len(method_labels)

    all_dfs = []

    for path, label in zip(load_paths, method_labels):
        print(f"[INFO] Loading: {label}\n       → {path}")

        if not os.path.exists(path):
            print(f"[WARN] File not found: {path}")
            continue

        L = np.load(path, allow_pickle=True)

        # Handle object dtype (SEA_AD)
        if L.dtype == object:
            L = np.array([np.array(row, dtype=float).ravel() for row in L])

        if L.ndim != 2:
            raise ValueError(f"[ERROR] Loading matrix must be 2D, got shape {L.shape}")

        h = compute_gene_entropy(L)
        df_tmp = pd.DataFrame({"Hoyer": h, "Method": label})
        all_dfs.append(df_tmp)

        print(f"    → {L.shape[0]} genes × {L.shape[1]} dims")

    df = pd.concat(all_dfs, ignore_index=True)
    palette = sns.color_palette("Set2", n_colors=len(df["Method"].unique()))

    plt.figure(figsize=(3.5, 2.5), dpi=200)

    # ---------------- Violin ----------------
    sns.violinplot(
        data=df, x="Method", y="Hoyer",
        palette=palette, inner=None,
        cut=0, linewidth=0.8,
        width=0.28
    )

    # ---------------- Transparent boxplot ----------------
    sns.boxplot(
        data=df, x="Method", y="Hoyer",
        width=0.18, showcaps=True,
        boxprops={"facecolor": (0, 0, 0, 0), "edgecolor": "black", "linewidth": 1.2},
        whiskerprops={"linewidth": 1.2},
        medianprops={"color": "black", "linewidth": 1.3},
        showfliers=False
    )

    # plt.title(title, fontsize=14, fontweight="bold", pad=15)
    plt.ylabel("Hoyer Sparsity", fontsize=11, fontweight="bold")
    plt.xlabel("Method", fontsize=11, fontweight="bold")

    plt.ylim(0.0, 1.0)
    plt.xticks(rotation=20, ha="right", fontsize=10)
    plt.tight_layout()

    # ---------------- Axis labels ----------------
    plt.xlabel("Method", fontsize=11, fontweight="bold")
    plt.ylabel("Hoyer Sparsity", fontsize=12, fontweight="bold")

    # ---------------- Y ticks ----------------
    yticks = np.arange(0.0, 1.01, 0.2)
    plt.yticks(yticks, [f"{y:.1f}" for y in yticks], fontsize=8, fontweight="bold")

    # ---------------- X tick labels ----------------
    plt.xticks(rotation=20, ha="right", fontsize=8, fontweight="bold")

    # ------------ Save PDF ------------
    save_path = os.path.join(FIGDIR, f"{save_name}.pdf")
    plt.savefig(save_path, dpi=300)
    print(f"[SAVED] → {save_path}")

    plt.show()


# 3. Su_2020
paths_su_2020 = [
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Su_2020/output/train_s_cd4_BL_2.0.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Su_2020/output/train_s_cd8_BL_2.0.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Su_2020/output_v2_4k/train_s_cd4_cd8_BL_Spectra_union.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Su_2020/output_v2/train_s_cd4_BL_sciRED.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Su_2020/output_v2/train_s_cd8_BL_sciRED.npy"
]

labels_su_2020 = [
    "GNN CD4",
    "GNN CD8",
    "Spectra",
    "sciRED CD4",
    "sciRED CD8",
]

plot_loading_sparsity(paths_su_2020, labels_su_2020,
                      "Loading Sparsity Comparison — Su 2020 (CD4/CD8)",
                      "Su2020_loading_sparsity")


# 4. SEA_AD
paths_sea_ad = [
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/SEA_AD/output/train_s_Astro_2.0.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/SEA_AD/output/train_s_Micro-PVM_2.0.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/SEA_AD/output_v2_4k/train_s_Astro_Micro-PVM_union.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/SEA_AD/output_v2/train_s_Astro_sciRED.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/SEA_AD/output_v2/train_s_Micro-PVM_sciRED.npy"
]

labels_sea_ad = [
    "GNN Astro",
    "GNN MicroPVM",
    "Spectra",
    "sciRED Astro",
    "sciRED MicroPVM",
]

plot_loading_sparsity(paths_sea_ad, labels_sea_ad,
                      "Loading Sparsity Comparison — SEA AD (Astro/Micro-PVM)",
                      "SEA_AD_loading_sparsity")


# # 5. Old Yost_2019
# paths_yost = [
#     "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Yost_2019/output/train_s_CD8T_2.0.npy",
#     "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/SF_2018/output_v2/train_s_CD8T_Spectra.npy",
#     # "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Yost_2019/output_v2_normalized/train_s_CD8T_Spectra.npy",
#     "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/SF_2018/output_v2/train_s_CD8T_sciRED.npy",
# ]

# 5. New Yost_2019
paths_yost = [
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Yost_2019/output_v4/output/train_s_CD8T_GNN_2.0.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Yost_2019/output_v4/train_s_CD8T_Spectra.npy",
    "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Yost_2019/output_v4/train_s_CD8T_sciRED.npy",
]

labels_yost = [
    "GNN",
    "Spectra",
    "sciRED"
]

plot_loading_sparsity(paths_yost, labels_yost,
                      "Loading Sparsity Comparison — Yost 2019 (CD8T)",
                      "Yost2019_loading_sparsity")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_auc_score
import os

root = "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Su_2020/"
fig_root = "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/"

fig_dir = os.path.join(fig_root, "Figures/Prediction_Scatterplots")
os.makedirs(fig_dir, exist_ok=True)

# ---- Su_2020 GNN (SDAN, weight=2.0) ----
gnn_cd4_path = os.path.join(root, "output/score_ind_cd4_BL_2.0.npy")
gnn_cd8_path = os.path.join(root, "output/score_ind_cd8_BL_2.0.npy")

# ---- Spectra (union) ----
spectra_cd4_csv = os.path.join(root, "output_v2_4k/results_classifiers_cd4_BL_Spectra_union_Spectra_union_logreg_ind_scores.csv")
spectra_cd8_csv = os.path.join(root, "output_v2_4k/results_classifiers_cd8_BL_Spectra_union_Spectra_union_logreg_ind_scores.csv")

# ---- sciRED ----
scired_cd4_csv = os.path.join(root, "output_v2/results_classifiers_cd4_BL_sciRED_sciRED_logreg_ind_scores.csv")
scired_cd8_csv = os.path.join(root, "output_v2/results_classifiers_cd8_BL_sciRED_sciRED_logreg_ind_scores.csv")


# LOADERS 
def load_npy_score(path):
    """Loads GNN numpy score file."""
    arr = np.load(path, allow_pickle=True)

    # Convert object array into numeric
    if arr.dtype == object:
        cleaned = []
        for row in arr:
            row = np.array(row, dtype=float).reshape(-1)
            cleaned.append(row[:2])
        arr = np.vstack(cleaned)

    label = arr[:, 0].astype(int)
    score = arr[:, 1].astype(float)

    return label, score


def load_csv_score(path):
    return pd.read_csv(path)


# Load everything 
label_cd4, sdan_cd4 = load_npy_score(gnn_cd4_path)
label_cd8, sdan_cd8 = load_npy_score(gnn_cd8_path)

spectra_cd4 = load_csv_score(spectra_cd4_csv)
spectra_cd8 = load_csv_score(spectra_cd8_csv)

scired_cd4 = load_csv_score(scired_cd4_csv)
scired_cd8 = load_csv_score(scired_cd8_csv)


def merge_scores(label, sdan, df_other, cell_type, other_name):
    """Align scores safely and create merged dataframe."""
    n = min(len(sdan), len(df_other))

    df = pd.DataFrame({
        "Label": np.where(label[:n] == 1, "Severe", "Mild"),
        "GNN": sdan[:n],
        other_name: df_other["score_pos"].values[:n],
        "CellType": cell_type,
    })
    return df


df_spec_cd4 = merge_scores(label_cd4, sdan_cd4, spectra_cd4, "CD4", "Spectra")
df_spec_cd8 = merge_scores(label_cd8, sdan_cd8, spectra_cd8, "CD8", "Spectra")
df_spec_all = pd.concat([df_spec_cd4, df_spec_cd8], ignore_index=True)

df_red_cd4 = merge_scores(label_cd4, sdan_cd4, scired_cd4, "CD4", "sciRED")
df_red_cd8 = merge_scores(label_cd8, sdan_cd8, scired_cd8, "CD8", "sciRED")
df_red_all = pd.concat([df_red_cd4, df_red_cd8], ignore_index=True)


def plot_scatter_by_celltype(df, other_name, title_suffix, save_path):
    """Scatterplot of GNN vs other method for CD4 and CD8 separately."""

    fig, axes = plt.subplots(1, 2, figsize=(7, 3), dpi=300, sharex=True, sharey=True)
    palette = {"Mild": "#1f77b4", "Severe": "#ff7f0e"}

    for i, ct in enumerate(["CD4", "CD8"]):

        ax = axes[i]
        sub = df[df["CellType"] == ct]

        if len(sub) == 0:
            ax.set_visible(False)
            continue

        # correlation + AUCs
        r = np.corrcoef(sub["GNN"], sub[other_name])[0, 1]
        auc_gnn = roc_auc_score((sub["Label"] == "Severe").astype(int), sub["GNN"])
        auc_other = roc_auc_score((sub["Label"] == "Severe").astype(int), sub[other_name])

        sns.scatterplot(
            data=sub, x="GNN", y=other_name,
            hue="Label", palette=palette,
            s=45, edgecolor="0.3", linewidth=0.6,
            alpha=0.9, ax=ax,
        )

        ax.plot([0, 1], [0, 1], "--", color="gray", linewidth=1)
        ax.set_title(ct, fontsize=11, fontweight="bold")

        ax.set_xlabel("GNN", fontsize=10, fontweight="bold")
        if i == 0:
            ax.set_ylabel(other_name, fontsize=10, fontweight="bold")
        else:
            ax.set_ylabel("")

        ax.set_xlim(0, 1)
        ax.set_ylim(0, 1)
        ax.grid(True, alpha=0.4)

        ax.tick_params(labelsize=9)
        for tick in ax.get_xticklabels(): tick.set_fontweight("bold")
        for tick in ax.get_yticklabels(): tick.set_fontweight("bold")

        # annotation
        ax.text(
            0.05, 0.95,
            f"Correlation (r) = {r:.2f}\nAUC (GNN) = {auc_gnn:.2f}\nAUC ({other_name}) = {auc_other:.2f}",
            transform=ax.transAxes,
            fontsize=7.5, va="top",
            bbox=dict(facecolor="white", alpha=0.8, edgecolor="none")
        )

        if i == 0:
            ax.legend(title="Label", fontsize=7, title_fontsize=7)
        else:
            ax.get_legend().remove()

    plt.suptitle(f"GNN vs {other_name} ({title_suffix})", fontsize=12, fontweight="bold")
    plt.tight_layout(rect=[0, 0, 1, 0.95])

    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"[SAVED] {save_path}")


# RUN PLOTS
plot_scatter_by_celltype(
    df_spec_all, "Spectra",
    "Su 2020", os.path.join(fig_dir, "GNN_vs_Spectra_Su2020_byCellType.pdf")
)

plot_scatter_by_celltype(
    df_red_all, "sciRED",
    "Su 2020", os.path.join(fig_dir, "GNN_vs_sciRED_Su2020_byCellType.pdf")
)
```

## 4. Biological Network Evaluation (Edge Quantiles)

```python
import os
import random
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from SDAN.preprocess import construct_gene_graph

# Reproducibility
SEED = 888
np.random.seed(SEED)
random.seed(SEED)
try:
    import torch
    torch.manual_seed(SEED)
    torch.cuda.manual_seed_all(SEED)
except:
    pass

# 1. Normalize rows for Spectra 
def normalize_rows_sum(x):
    """Normalize each row so values sum to 1 (rowwise probability distribution)."""
    x = np.asarray(x, float)
    row_sum = x.sum(axis=1, keepdims=True)
    return x / row_sum

# 2. Biological connectivity quantile
def compute_connectivity_uniform_fast(train_s, gene_list, edge_list_index, cutoff=0.8):
    """
    Compute:
      - component size (#genes > cutoff)
      - biological edge degree (from BioGRID)
      - quantile vs null random modules
    """

    train_s = train_s.T  # components × genes

    # Build adjacency matrix
    n = len(gene_list)
    A = np.zeros((n, n), dtype=np.uint8)
    A[edge_list_index[0], edge_list_index[1]] = 1
    A[edge_list_index[1], edge_list_index[0]] = 1

    size_list = []
    degree_list = []

    # Component connectivity
    for row in train_s:
        genes = np.where(row > cutoff)[0]
        k = len(genes)
        if k == 0:
            continue

        subA = A[genes][:, genes]
        degree_list.append(subA.sum() / k)
        size_list.append(k)

    size_list = np.array(size_list)
    degree_list = np.array(degree_list)

    # Null distribution (random gene sets)
    size_unique = np.unique(size_list)
    degree_sim = {}

    for sz in size_unique:
        sims = []
        for _ in range(1000):
            idx = np.random.choice(n, sz, replace=False)
            subA = A[idx][:, idx]
            sims.append(subA.sum() / sz)
        degree_sim[sz] = np.array(sims)

    # Quantile vs null
    quantile = []
    for sz, deg in zip(size_list, degree_list):
        sim = degree_sim[sz]
        quantile.append((sim < deg).mean())

    return size_list, degree_list, np.array(quantile)

COLOR_MAP = {
    # GNN category 1
    "GNN CD4": "#1f77b4",        # blue
    "GNN Astro": "#1f77b4",      # blue
    "GNN CD8T": "#1f77b4",        # blue   

    # GNN category 2
    "GNN CD8": "#ff7f0e",        # orange
    "GNN MicroPVM": "#ff7f0e",   # orange

    # Spectra (always single category)
    "Spectra": "#2ca02c",                # green
    "Spectra CD8T": "#2ca02c",            # green
    "Spectra Astro-MicroPVM": "#2ca02c",  # green
}

# 3. Final SDAN-paper style violin + jitter plot
def plot_quantile_edge_violin(quantile_edge_dict, save_path, title):
    """
    - Matplotlib violinplot (medians shown)
    - tab10 colors (gray for empty)
    - black jittered points
    - bold labels, 0–1 y-range
    """

    labels = list(quantile_edge_dict.keys())
    values = [v if len(v) > 0 else np.zeros(1) for v in quantile_edge_dict.values()]

    plt.figure(figsize=(5, 4), dpi=300)

    # Violin plot (matplotlib)
    parts = plt.violinplot(
        values,
        positions=range(len(labels)),
        showmeans=False,
        showmedians=True,
    )

    # Color each violin
    # for i, v in enumerate(values):
    #     color = "gray" if (len(v) == 1 and v[0] == 0) else plt.cm.tab10(i)
    #     body = parts["bodies"][i]
    #     body.set_facecolor(color)
    #     body.set_alpha(0.75)
    #     body.set_edgecolor("black")
    #     body.set_linewidth(1.0)

    for i, (label, v) in enumerate(zip(labels, values)):
        color = COLOR_MAP.get(label, "gray")

        body = parts["bodies"][i]
        body.set_facecolor(color)
        body.set_alpha(0.75)
        body.set_edgecolor("black")
        body.set_linewidth(1.0)

    # Add jittered dot plot
    for i, v in enumerate(values):
        sns.stripplot(
            x=[i] * len(v),
            y=v,
            color="black",
            size=3,
            alpha=0.65,
            jitter=0.18
        )

    # Labels and formatting
    # plt.title(title, fontsize=13, fontweight="bold", pad=10)
    plt.xlabel("Method", fontsize=11, fontweight="bold")
    plt.ylabel("Quantile of edges per component", fontsize=11, fontweight="bold")

    plt.xticks(range(len(labels)), labels, rotation=15, fontsize=10, fontweight="bold")
    plt.yticks(fontsize=10, fontweight="bold")

    plt.ylim(0, 1.05)
    plt.grid(True, alpha=0.35)
    plt.tight_layout()

    # Save as PDF (always)
    if not save_path.endswith(".pdf"):
        save_path = save_path.replace(".png", ".pdf")

    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"[SAVED] {save_path}")

# 4. Run on Su_2020 GNN vs Spectra (CD4, CD8, Union)
base = "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Su_2020/"

paths = {
    "GNN CD4": {
        "load": f"{base}output/train_s_cd4_BL_2.0.npy",
        "genes": f"{base}output/gene_list_cd4_BL_2.0.txt",
        "type": "sdan",
    },
    "GNN CD8": {
        "load": f"{base}output/train_s_cd8_BL_2.0.npy",
        "genes": f"{base}output/gene_list_cd8_BL_2.0.txt",
        "type": "sdan",
    },
    "Spectra": {
        "load": f"{base}output_v2_4k/train_s_cd4_cd8_BL_Spectra_union.npy",
        "genes": f"{base}output_v2_4k/gene_list_cd4_cd8_union.txt",
        "type": "spectra",
    },
}

results = {}

for name, cfg in paths.items():
    print(f"\n[INFO] Processing {name}")

    train_s = np.load(cfg["load"])

    # Load gene list
    with open(cfg["genes"], "r") as f:
        genes = pd.Index([line.strip().split()[0] for line in f if line.strip()])

    # Normalize Spectra
    if cfg["type"] == "spectra":
        print("  → Normalizing Spectra (row-sum)")
        train_s = normalize_rows_sum(train_s)

    # BioGRID graph
    edge_list_index = construct_gene_graph(genes).detach().numpy()

    # Compute connectivity
    size, degree, quantile = compute_connectivity_uniform_fast(
        train_s, genes, edge_list_index
    )

    results[name] = {"size": size, "degree": degree, "quantile": quantile}

print("\n[DONE] Connectivity computation complete.")

# 5. Plot all quantiles together
quantile_dict = {name: results[name]["quantile"] for name in results}

fig_base = "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/"
out_path = f"{fig_base}Figures/BioGRID Connectivity/quantile_edge_GNN_vs_Spectra.pdf"

plot_quantile_edge_violin(
    quantile_dict,
    save_path=out_path,
    title="BioGRID Connectivity — GNN vs Spectra"
)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

BASE = "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN/Yost_2019"
FIGS_BASE = "/Users/yuanyuangao/Library/Mobile Documents/com~apple~CloudDocs/Research/SDAN"

OUT_FIGS = os.path.join(FIGS_BASE, "Figures/Boxplot of Predicted Severity Scores/")
os.makedirs(OUT_FIGS, exist_ok=True)

# # Old
# paths = {
#     "CD8T (sciRED)": f"{BASE}/output_v2/results_classifiers_CD8T_sciRED_sciRED_logreg_ind_scores.csv",
#     "CD8T (GNN)":    f"{BASE}/output/score_ind_CD8T_2.0.npy",
#     "CD8T (Spectra)":f"{BASE}/output_v2/results_classifiers_CD8T_Spectra_Spectra_logreg_ind_scores.csv",
# }

# New
paths = {
    "CD8T (sciRED)": f"{BASE}/output_v4/results_classifiers_CD8T_sciRED_sciRED_logreg_ind_scores.csv",
    "CD8T (GNN)":    f"{BASE}/output_v4/results_classifiers_CD8T_GNN_2.0_GNN_logreg_ind_scores.csv",
    "CD8T (Spectra)":f"{BASE}/output_v4/results_classifiers_CD8T_Spectra_Spectra_logreg_ind_scores.csv",
}

# Colors
palette = sns.color_palette("coolwarm", 2)
COLOR_MAP = {"No": palette[0], "Yes": palette[1]}
GROUP_ORDER = ["No", "Yes"]

# Load NPY
def load_npy_as_df(path):
    """
    Handles Yost .npy arrays:
        shape (N,2) or object array [label, score, ...]
    """
    raw = np.load(path, allow_pickle=True)

    if raw.dtype == object:
        cleaned = []
        for row in raw:
            row = np.array(row, dtype=float).reshape(-1)
            cleaned.append(row[:2])
        raw = np.vstack(cleaned)

    labels = raw[:, 0].astype(int)
    scores  = raw[:, 1].astype(float)

    label_map = {0: "No", 1: "Yes"}

    return pd.DataFrame({
        "label_str": [label_map[x] for x in labels],
        "score_pos": scores,
    })

def plot_boxplot(df, title):
    fig, ax = plt.subplots(figsize=(3, 3), dpi=300)

    sns.boxplot(
        data=df, x="label_str", y="score_pos",
        order=GROUP_ORDER, palette=COLOR_MAP,
        width=0.25, linewidth=1.2, showfliers=False,
        boxprops=dict(alpha=0.75),
        ax=ax
    )

    sns.stripplot(
        data=df, x="label_str", y="score_pos",
        order=GROUP_ORDER, palette=COLOR_MAP,
        size=3, jitter=0.15, alpha=0.75,
        linewidth=0.3, edgecolor="black",
        ax=ax
    )

    ax.set_title(title, fontsize=11, fontweight="bold")
    ax.set_xlabel("Clinical Group", fontsize=11, fontweight="bold")
    ax.set_ylabel("Predicted Probability", fontsize=11, fontweight="bold")
    ax.set_ylim(0, 1)
    ax.tick_params(labelsize=10)

    for t in ax.get_xticklabels():
        t.set_fontweight("bold")
    for t in ax.get_yticklabels():
        t.set_fontweight("bold")

    plt.tight_layout()
    out = os.path.join(OUT_FIGS, f"Yost_boxplot_{title.replace(' ', '_')}.pdf")
    plt.savefig(out, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close(fig)

    print(f"[SAVED] {out}")

def plot_all_combined(dfs_dict):
    ordered_tags = ["CD8T (sciRED)", "CD8T (GNN)", "CD8T (Spectra)"]

    fig, axes = plt.subplots(1, 3, figsize=(9, 3.4), dpi=300)

    for ax, tag in zip(axes, ordered_tags):
        df = dfs_dict[tag]

        sns.boxplot(
            data=df, x="label_str", y="score_pos",
            order=GROUP_ORDER, palette=COLOR_MAP,
            width=0.25, linewidth=1.2, showfliers=False,
            boxprops=dict(alpha=0.75), ax=ax
        )

        sns.stripplot(
            data=df, x="label_str", y="score_pos",
            order=GROUP_ORDER, palette=COLOR_MAP,
            size=3, jitter=0.15, alpha=0.75,
            linewidth=0.3, edgecolor="black", ax=ax
        )

        ax.set_title(tag, fontsize=10, fontweight="bold")
        ax.set_xlabel("")   # shared labels added later
        ax.set_ylabel("")
        ax.set_ylim(0, 1)
        ax.tick_params(labelsize=9)

    # Shared axis labels
    fig.supxlabel("Clinical Group", fontsize=13, fontweight="bold", y=0.08)
    fig.supylabel("Predicted Probability", fontsize=13, fontweight="bold", x=0.04)

    # Main title
    fig.suptitle(
        "CD8T Response Probability Across Methods — Yost 2019",
        fontsize=15, fontweight="bold", y=0.98
    )

    plt.tight_layout(rect=[0.05, 0.12, 1, 0.92])
    out = os.path.join(OUT_FIGS, "Yost_combined_boxplot.pdf")
    plt.savefig(out, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close(fig)

    print(f"[SAVED] {out}")

# Main
dfs_dict = {}

for tag, path in paths.items():
    print(f"[LOAD] {tag}: {path}")

    if path.endswith(".npy"):
        df = load_npy_as_df(path)
    else:
        df = pd.read_csv(path)
        df["label_str"] = df["label_str"].replace({"R": "Yes", "NR": "No"})

    dfs_dict[tag] = df

    # Individual plot
    plot_boxplot(df, tag)

# Combined panel
plot_all_combined(dfs_dict)
```
