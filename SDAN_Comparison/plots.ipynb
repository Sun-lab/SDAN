## 1. Entropy (Hoyer Sparsity) Violin Plot for Loadings

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
from pathlib import Path
from scipy.cluster.hierarchy import linkage, leaves_list


def dataset_figure_dir(path):
    path = Path(path)
    dataset = next(part for part in path.parts if part in {"Su_2020", "SEA_AD", "Yost_2019"})
    figdir = Path("SDAN_Comparison") / dataset / "figures"
    figdir.mkdir(parents=True, exist_ok=True)
    return figdir


def load_matrix(path):
    arr = np.load(path, allow_pickle=True)
    if arr.dtype == object:
        arr = np.array([np.asarray(row, dtype=float).ravel() for row in arr])
    if arr.ndim == 1:
        arr = arr.reshape(1, -1)
    return arr


def plot_clustermaps(named_files, cmap="Blues", col_cluster=False, row_cluster=True):
    shapes = {}

    for name, path in named_files.items():
        print(f"\n=== Loading [{name}] ===")
        print(path)

        arr = load_matrix(path)
        shapes[name] = arr.shape
        print(f"Shape: {arr.shape}")

        g = sns.clustermap(
            arr,
            col_cluster=col_cluster,
            row_cluster=row_cluster,
            cmap=cmap,
            figsize=(14, 6)
        )

        plt.suptitle(f"{name}", fontsize=18, fontweight="bold")
        g.ax_heatmap.set_xlabel("40 gene programs", fontsize=15, fontweight="bold")
        g.ax_heatmap.set_ylabel(f"{arr.shape[0]} genes", fontsize=15, fontweight="bold")
        g.ax_heatmap.set_xticks([])
        g.ax_heatmap.set_yticks([])

        save_path = dataset_figure_dir(path) / f"{name.replace(' ', '_')}.pdf"
        g.fig.savefig(save_path, bbox_inches="tight")
        print(f"[SAVED] {save_path}")
        plt.show()
        plt.close(g.fig)

    return shapes


def clustered_array(arr, col_cluster=False, row_cluster=True):
    row_order = np.arange(arr.shape[0])
    col_order = np.arange(arr.shape[1])

    if row_cluster and arr.shape[0] > 1:
        row_linkage = linkage(arr, method="average", metric="euclidean")
        row_order = leaves_list(row_linkage)

    if col_cluster and arr.shape[1] > 1:
        col_linkage = linkage(arr.T, method="average", metric="euclidean")
        col_order = leaves_list(col_linkage)

    return arr[np.ix_(row_order, col_order)]


def plot_two_panel_clustermaps(dataset, panel_paths, save_name, cmap="Blues", col_cluster=False, row_cluster=True):
    shapes = {}
    arrays = {}
    clustered_arrays = {}
    vmin = None
    vmax = None

    for name, path in panel_paths.items():
        print(f"\n=== Loading [{name}] ===")
        print(path)
        arr = load_matrix(path)
        arrays[name] = arr
        shapes[name] = arr.shape
        print(f"Shape: {arr.shape}")
        clustered = clustered_array(arr, col_cluster=col_cluster, row_cluster=row_cluster)
        clustered_arrays[name] = clustered
        arr_min = float(np.min(clustered))
        arr_max = float(np.max(clustered))
        vmin = arr_min if vmin is None else min(vmin, arr_min)
        vmax = arr_max if vmax is None else max(vmax, arr_max)

    fig, axes = plt.subplots(1, 2, figsize=(16, 6), dpi=200)
    cbar_ax = fig.add_axes([0.92, 0.18, 0.015, 0.64])

    for i, (name, arr) in enumerate(clustered_arrays.items()):
        sns.heatmap(
            arr,
            cmap=cmap,
            ax=axes[i],
            cbar=(i == len(clustered_arrays) - 1),
            cbar_ax=cbar_ax if i == len(clustered_arrays) - 1 else None,
            xticklabels=False,
            yticklabels=False,
            vmin=vmin,
            vmax=vmax,
        )
        axes[i].set_title(name, fontsize=17, fontweight="bold")
        axes[i].set_xlabel("40 gene programs", fontsize=15, fontweight="bold")
        axes[i].set_ylabel(f"{arrays[name].shape[0]} genes", fontsize=15, fontweight="bold")
        axes[i].set_xticks([])
        axes[i].set_yticks([])

    fig.tight_layout(rect=[0, 0, 0.9, 1])
    save_path = Path("SDAN_Comparison") / dataset / "figures" / f"{save_name}.pdf"
    fig.savefig(save_path, bbox_inches="tight")
    print(f"[SAVED] {save_path}")
    plt.show()
    plt.close(fig)

    return shapes


single_files = {
    "Spectra CD4+CD8": os.path.join("SDAN_Comparison", "Su_2020", "output_comparison", "Spectra", "train_s_cd4_cd8_BL_Spectra_union.npy"),
    "Spectra Astro+MicroPVM": os.path.join("SDAN_Comparison", "SEA_AD", "output_comparison", "Spectra", "train_s_Astro_Micro-PVM_union.npy"),
    "Spectra CD8T": os.path.join("SDAN_Comparison", "Yost_2019", "output_comparison", "Spectra", "train_s_CD8T_Spectra.npy"),
    "sciRED CD8T": os.path.join("SDAN_Comparison", "Yost_2019", "output_comparison", "sciRED", "train_s_CD8T_sciRED.npy"),
    "SDAN CD8T": os.path.join("Yost_2019", "output", "train_s_CD8T_2.0.npy"),
}

shapes = plot_clustermaps(single_files)
shapes.update(
    plot_two_panel_clustermaps(
        "Su_2020",
        {
            "SDAN CD4": os.path.join("Su_2020", "output", "train_s_cd4_BL_2.0.npy"),
            "SDAN CD8": os.path.join("Su_2020", "output", "train_s_cd8_BL_2.0.npy"),
        },
        "SDAN_Su_2020_weight_2_1x2",
    )
)
shapes.update(
    plot_two_panel_clustermaps(
        "Su_2020",
        {
            "sciRED CD4": os.path.join("SDAN_Comparison", "Su_2020", "output_comparison", "sciRED", "train_s_cd4_BL_sciRED.npy"),
            "sciRED CD8": os.path.join("SDAN_Comparison", "Su_2020", "output_comparison", "sciRED", "train_s_cd8_BL_sciRED.npy"),
        },
        "sciRED_Su_2020_1x2",
    )
)
shapes.update(
    plot_two_panel_clustermaps(
        "SEA_AD",
        {
            "SDAN Astro": os.path.join("SEA_AD", "output", "train_s_Astro_2.0.npy"),
            "SDAN MicroPVM": os.path.join("SEA_AD", "output", "train_s_Micro-PVM_2.0.npy"),
        },
        "SDAN_SEA_AD_weight_2_1x2",
    )
)
shapes.update(
    plot_two_panel_clustermaps(
        "SEA_AD",
        {
            "sciRED Astro": os.path.join("SDAN_Comparison", "SEA_AD", "output_comparison", "sciRED", "train_s_Astro_sciRED.npy"),
            "sciRED MicroPVM": os.path.join("SDAN_Comparison", "SEA_AD", "output_comparison", "sciRED", "train_s_Micro-PVM_sciRED.npy"),
        },
        "sciRED_SEA_AD_1x2",
    )
)

print("\n=== All Array Shapes ===")
for name, shape in shapes.items():
    print(f"{name:25s} : {shape}")

```

## 2. Loading Sparsity Comparison (Su_2020, SEA_AD, Yost_2019)

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os


def ensure_figdir(dataset):
    figdir = os.path.join("SDAN_Comparison", dataset, "figures")
    os.makedirs(figdir, exist_ok=True)
    return figdir


def hoyer_sparsity(v):
    v = np.abs(v)
    return (np.sqrt(len(v)) - np.linalg.norm(v, 1) / np.linalg.norm(v, 2)) / (np.sqrt(len(v)) - 1 + 1e-12)


def compute_gene_entropy(L):
    return np.array([hoyer_sparsity(L[i, :]) for i in range(L.shape[0])])


def plot_loading_sparsity(load_paths, method_labels, save_name, dataset):
    sns.set_theme(style="whitegrid")
    assert len(load_paths) == len(method_labels)

    all_dfs = []

    for path, label in zip(load_paths, method_labels):
        print(f"[INFO] Loading: {label}\n       → {path}")

        if not os.path.exists(path):
            print(f"[WARN] File not found: {path}")
            continue

        L = np.load(path, allow_pickle=True)
        if L.dtype == object:
            L = np.array([np.array(row, dtype=float).ravel() for row in L])

        if L.ndim != 2:
            raise ValueError(f"[ERROR] Loading matrix must be 2D, got shape {L.shape}")

        h = compute_gene_entropy(L)
        all_dfs.append(pd.DataFrame({"Hoyer": h, "Method": label}))
        print(f"    → {L.shape[0]} genes × {L.shape[1]} dims")

    df = pd.concat(all_dfs, ignore_index=True)
    palette = sns.color_palette("Set2", n_colors=len(df["Method"].unique()))

    plt.figure(figsize=(3.5, 2.5), dpi=200)
    sns.violinplot(data=df, x="Method", y="Hoyer", palette=palette, inner=None, cut=0, linewidth=0.8, width=0.28)
    sns.boxplot(
        data=df,
        x="Method",
        y="Hoyer",
        width=0.18,
        showcaps=True,
        boxprops={"facecolor": (0, 0, 0, 0), "edgecolor": "black", "linewidth": 1.2},
        whiskerprops={"linewidth": 1.2},
        medianprops={"color": "black", "linewidth": 1.3},
        showfliers=False,
    )

    plt.xlabel("Method", fontsize=11, fontweight="bold")
    plt.ylabel("Hoyer Sparsity", fontsize=12, fontweight="bold")
    plt.ylim(0.0, 1.0)
    yticks = np.arange(0.0, 1.01, 0.2)
    plt.yticks(yticks, [f"{y:.1f}" for y in yticks], fontsize=8, fontweight="bold")
    plt.xticks(rotation=20, ha="right", fontsize=8, fontweight="bold")
    plt.tight_layout()

    save_path = os.path.join(ensure_figdir(dataset), f"{save_name}.pdf")
    plt.savefig(save_path, dpi=300)
    print(f"[SAVED] → {save_path}")
    plt.show()


paths_su_2020 = [
    os.path.join("SDAN_Comparison", "Su_2020", "output_comparison", "Spectra", "train_s_cd4_cd8_BL_Spectra_union.npy"),
    os.path.join("SDAN_Comparison", "Su_2020", "output_comparison", "sciRED", "train_s_cd4_BL_sciRED.npy"),
    os.path.join("SDAN_Comparison", "Su_2020", "output_comparison", "sciRED", "train_s_cd8_BL_sciRED.npy"),
]
labels_su_2020 = ["Spectra", "sciRED CD4", "sciRED CD8"]
plot_loading_sparsity(paths_su_2020, labels_su_2020, "Su2020_loading_sparsity", "Su_2020")

paths_sea_ad = [
    os.path.join("SDAN_Comparison", "SEA_AD", "output_comparison", "Spectra", "train_s_Astro_Micro-PVM_union.npy"),
    os.path.join("SDAN_Comparison", "SEA_AD", "output_comparison", "sciRED", "train_s_Astro_sciRED.npy"),
    os.path.join("SDAN_Comparison", "SEA_AD", "output_comparison", "sciRED", "train_s_Micro-PVM_sciRED.npy"),
]
labels_sea_ad = ["Spectra", "sciRED Astro", "sciRED MicroPVM"]
plot_loading_sparsity(paths_sea_ad, labels_sea_ad, "SEA_AD_loading_sparsity", "SEA_AD")

paths_yost = [
    os.path.join("SDAN_Comparison", "Yost_2019", "output_comparison", "Spectra", "train_s_CD8T_Spectra.npy"),
    os.path.join("SDAN_Comparison", "Yost_2019", "output_comparison", "sciRED", "train_s_CD8T_sciRED.npy"),
]
labels_yost = ["Spectra", "sciRED"]
plot_loading_sparsity(paths_yost, labels_yost, "Yost2019_loading_sparsity", "Yost_2019")

```

## 3. Scatterplot of Predictions (Individual-Level AUC Comparison)

### 3.1 Su 2020

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_auc_score
import os

root = os.path.join("SDAN_Comparison", "Su_2020", "output_comparison")
fig_dir = os.path.join("SDAN_Comparison", "Su_2020", "figures")
os.makedirs(fig_dir, exist_ok=True)

sdan_cd4_path = os.path.join(root, "SDAN", "results_classifiers_cd4_BL_SDAN_2.0_logreg_individual_scores.csv")
sdan_cd8_path = os.path.join(root, "SDAN", "results_classifiers_cd8_BL_SDAN_2.0_logreg_individual_scores.csv")
spectra_cd4_csv = os.path.join(root, "Spectra", "results_classifiers_cd4_BL_Spectra_union_Spectra_union_logreg_ind_scores.csv")
spectra_cd8_csv = os.path.join(root, "Spectra", "results_classifiers_cd8_BL_Spectra_union_Spectra_union_logreg_ind_scores.csv")
scired_cd4_csv = os.path.join(root, "sciRED", "results_classifiers_cd4_BL_sciRED_logreg_individual_scores.csv")
scired_cd8_csv = os.path.join(root, "sciRED", "results_classifiers_cd8_BL_sciRED_logreg_individual_scores.csv")


def load_score_csv(path, positive_label, negative_label):
    df = pd.read_csv(path)
    if "label_str" in df.columns:
        df["label_str"] = df["label_str"].replace({negative_label.lower(): negative_label, positive_label.lower(): positive_label})
        df["label_str"] = df["label_str"].str.capitalize()
    return df


sdan_cd4 = load_score_csv(sdan_cd4_path, "Severe", "Mild")
sdan_cd8 = load_score_csv(sdan_cd8_path, "Severe", "Mild")
spectra_cd4 = load_score_csv(spectra_cd4_csv, "Severe", "Mild")
spectra_cd8 = load_score_csv(spectra_cd8_csv, "Severe", "Mild")
scired_cd4 = load_score_csv(scired_cd4_csv, "Severe", "Mild")
scired_cd8 = load_score_csv(scired_cd8_csv, "Severe", "Mild")


def merge_scores(df_sdan, df_other, cell_type, other_name):
    n = min(len(df_sdan), len(df_other))
    return pd.DataFrame({
        "Label": df_sdan["label_str"].values[:n],
        "SDAN": df_sdan["score_pos"].values[:n],
        other_name: df_other["score_pos"].values[:n],
        "CellType": cell_type,
    })


df_spec_all = pd.concat([
    merge_scores(sdan_cd4, spectra_cd4, "CD4", "Spectra"),
    merge_scores(sdan_cd8, spectra_cd8, "CD8", "Spectra"),
], ignore_index=True)

df_red_all = pd.concat([
    merge_scores(sdan_cd4, scired_cd4, "CD4", "sciRED"),
    merge_scores(sdan_cd8, scired_cd8, "CD8", "sciRED"),
], ignore_index=True)


def plot_scatter_by_celltype(df, other_name, title_suffix, save_path):
    fig, axes = plt.subplots(1, 2, figsize=(7, 3), dpi=300, sharex=True, sharey=True)
    palette = {"Mild": "#1f77b4", "Severe": "#ff7f0e"}

    for i, ct in enumerate(["CD4", "CD8"]):
        ax = axes[i]
        sub = df[df["CellType"] == ct]
        if len(sub) == 0:
            ax.set_visible(False)
            continue

        r = np.corrcoef(sub["SDAN"], sub[other_name])[0, 1]
        auc_sdan = roc_auc_score((sub["Label"] == "Severe").astype(int), sub["SDAN"])
        auc_other = roc_auc_score((sub["Label"] == "Severe").astype(int), sub[other_name])

        sns.scatterplot(data=sub, x="SDAN", y=other_name, hue="Label", palette=palette, s=45, edgecolor="0.3", linewidth=0.6, alpha=0.9, ax=ax)
        ax.plot([0, 1], [0, 1], "--", color="gray", linewidth=1)
        ax.set_title(ct, fontsize=11, fontweight="bold")
        ax.set_xlabel("SDAN", fontsize=10, fontweight="bold")
        ax.set_ylabel(other_name if i == 0 else "", fontsize=10, fontweight="bold")
        ax.set_xlim(0, 1)
        ax.set_ylim(0, 1)
        ax.grid(True, alpha=0.4)
        ax.tick_params(labelsize=9)

        ax.text(
            0.05,
            0.95,
            f"Correlation (r) = {r:.2f}\nAUC (SDAN) = {auc_sdan:.2f}\nAUC ({other_name}) = {auc_other:.2f}",
            transform=ax.transAxes,
            fontsize=7.5,
            va="top",
            bbox=dict(facecolor="white", alpha=0.8, edgecolor="none"),
        )

        if i == 0:
            ax.legend(title="Label", fontsize=7, title_fontsize=7)
        else:
            ax.get_legend().remove()

    plt.suptitle(f"SDAN vs {other_name} ({title_suffix})", fontsize=12, fontweight="bold")
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"[SAVED] {save_path}")


plot_scatter_by_celltype(df_spec_all, "Spectra", "Su 2020", os.path.join(fig_dir, "SDAN_vs_Spectra_Su2020_byCellType.pdf"))
plot_scatter_by_celltype(df_red_all, "sciRED", "Su 2020", os.path.join(fig_dir, "SDAN_vs_sciRED_Su2020_byCellType.pdf"))

```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_auc_score

root = os.path.join("SDAN_Comparison", "SEA_AD", "output_comparison")
fig_dir = os.path.join("SDAN_Comparison", "SEA_AD", "figures")
os.makedirs(fig_dir, exist_ok=True)

sdan_astro_path = os.path.join(root, "SDAN", "results_classifiers_Astro_SDAN_2.0_logreg_individual_scores.csv")
sdan_micro_path = os.path.join(root, "SDAN", "results_classifiers_Micro-PVM_SDAN_2.0_logreg_individual_scores.csv")
spectra_astro_csv = os.path.join(root, "Spectra", "results_classifiers_Astro_Spectra_union_Spectra_union_logreg_ind_scores.csv")
spectra_micro_csv = os.path.join(root, "Spectra", "results_classifiers_Micro-PVM_Spectra_union_Spectra_union_logreg_ind_scores.csv")
scired_astro_csv = os.path.join(root, "sciRED", "results_classifiers_Astro_sciRED_logreg_individual_scores.csv")
scired_micro_csv = os.path.join(root, "sciRED", "results_classifiers_Micro-PVM_sciRED_logreg_individual_scores.csv")


def load_score_csv(path):
    df = pd.read_csv(path)
    if "label_str" in df.columns:
        df["label_str"] = df["label_str"].replace({"No dementia": "No dementia", "Dementia": "Dementia"})
    return df


sdan_astro = load_score_csv(sdan_astro_path)
sdan_micro = load_score_csv(sdan_micro_path)
spectra_astro = load_score_csv(spectra_astro_csv)
spectra_micro = load_score_csv(spectra_micro_csv)
scired_astro = load_score_csv(scired_astro_csv)
scired_micro = load_score_csv(scired_micro_csv)


def merge_scores(df_sdan, df_other, cell_type, other_name):
    n = min(len(df_sdan), len(df_other))
    return pd.DataFrame({
        "Label": df_sdan["label_str"].values[:n],
        "SDAN": df_sdan["score_pos"].values[:n],
        other_name: df_other["score_pos"].values[:n],
        "CellType": cell_type,
    })


df_spec_all = pd.concat([
    merge_scores(sdan_astro, spectra_astro, "Astro", "Spectra"),
    merge_scores(sdan_micro, spectra_micro, "Micro-PVM", "Spectra"),
], ignore_index=True)

df_red_all = pd.concat([
    merge_scores(sdan_astro, scired_astro, "Astro", "sciRED"),
    merge_scores(sdan_micro, scired_micro, "Micro-PVM", "sciRED"),
], ignore_index=True)


def plot_scatter_by_celltype(df, other_name, title_suffix, save_path):
    fig, axes = plt.subplots(1, 2, figsize=(7, 3), dpi=300, sharex=True, sharey=True)
    palette = {"No dementia": "#1f77b4", "Dementia": "#ff7f0e"}

    for i, ct in enumerate(["Astro", "Micro-PVM"]):
        ax = axes[i]
        sub = df[df["CellType"] == ct].copy()
        if len(sub) == 0:
            ax.set_visible(False)
            continue

        r = np.corrcoef(sub["SDAN"], sub[other_name])[0, 1]
        auc_sdan = roc_auc_score((sub["Label"] == "Dementia").astype(int), sub["SDAN"])
        auc_other = roc_auc_score((sub["Label"] == "Dementia").astype(int), sub[other_name])

        sns.scatterplot(data=sub, x="SDAN", y=other_name, hue="Label", palette=palette, s=45, edgecolor="0.3", linewidth=0.6, alpha=0.9, ax=ax)
        ax.plot([0, 1], [0, 1], "--", color="gray", linewidth=1)
        ax.set_title(ct, fontsize=11, fontweight="bold")
        ax.set_xlabel("SDAN", fontsize=10, fontweight="bold")
        ax.set_ylabel(other_name if i == 0 else "", fontsize=10, fontweight="bold")
        ax.set_xlim(0, 1)
        ax.set_ylim(0, 1)
        ax.grid(True, alpha=0.35)

        ticks = np.arange(0.0, 1.01, 0.2)
        ax.set_xticks(ticks)
        ax.set_yticks(ticks)
        ax.set_xticklabels([f"{t:.1f}" for t in ticks], fontsize=9, fontweight="bold")
        ax.set_yticklabels([f"{t:.1f}" for t in ticks], fontsize=9, fontweight="bold")

        ax.text(
            0.04,
            0.96,
            f"Correlation (r) = {r:.2f}\nAUC (SDAN) = {auc_sdan:.2f}\nAUC ({other_name}) = {auc_other:.2f}",
            transform=ax.transAxes,
            fontsize=7.5,
            va="top",
            ha="left",
            bbox=dict(facecolor="white", alpha=0.85, edgecolor="0.7", boxstyle="round,pad=0.2"),
        )

        if i == 0:
            ax.legend(title="Label", fontsize=7, title_fontsize=7)
        else:
            ax.get_legend().remove()

    plt.suptitle(f"SDAN vs {other_name} ({title_suffix})", fontsize=12, fontweight="bold")
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"[SAVED] {save_path}")


plot_scatter_by_celltype(df_spec_all, "Spectra", "SEA AD", os.path.join(fig_dir, "SDAN_vs_Spectra_SEA_AD_by_CellType.pdf"))
plot_scatter_by_celltype(df_red_all, "sciRED", "SEA AD", os.path.join(fig_dir, "SDAN_vs_sciRED_SEA_AD_by_CellType.pdf"))

```

## 4. Biological Network Evaluation (Edge Quantiles)

```python
import os
import random
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from SDAN.preprocess import construct_gene_graph

SEED = 888
np.random.seed(SEED)
random.seed(SEED)
try:
    import torch
    torch.manual_seed(SEED)
    torch.cuda.manual_seed_all(SEED)
except:
    pass


def normalize_rows_sum(x):
    x = np.asarray(x, float)
    row_sum = x.sum(axis=1, keepdims=True)
    return x / row_sum


def compute_connectivity_uniform_fast(train_s, gene_list, edge_list_index, cutoff=0.8):
    train_s = train_s.T
    n = len(gene_list)
    A = np.zeros((n, n), dtype=np.uint8)
    A[edge_list_index[0], edge_list_index[1]] = 1
    A[edge_list_index[1], edge_list_index[0]] = 1

    size_list = []
    degree_list = []
    for row in train_s:
        genes = np.where(row > cutoff)[0]
        k = len(genes)
        if k == 0:
            continue
        subA = A[genes][:, genes]
        degree_list.append(subA.sum() / k)
        size_list.append(k)

    size_list = np.array(size_list)
    degree_list = np.array(degree_list)
    size_unique = np.unique(size_list)
    degree_sim = {}

    for sz in size_unique:
        sims = []
        for _ in range(1000):
            idx = np.random.choice(n, sz, replace=False)
            subA = A[idx][:, idx]
            sims.append(subA.sum() / sz)
        degree_sim[sz] = np.array(sims)

    quantile = []
    for sz, deg in zip(size_list, degree_list):
        sim = degree_sim[sz]
        quantile.append((sim < deg).mean())

    return size_list, degree_list, np.array(quantile)


COLOR_MAP = {"Spectra": "#2ca02c", "sciRED CD4": "#e377c2", "sciRED CD8": "#bcbd22"}


def plot_quantile_edge_violin(quantile_edge_dict, save_path, title):
    labels = list(quantile_edge_dict.keys())
    values = [v if len(v) > 0 else np.zeros(1) for v in quantile_edge_dict.values()]
    plt.figure(figsize=(5, 4), dpi=300)
    parts = plt.violinplot(values, positions=range(len(labels)), showmeans=False, showmedians=True)

    for i, (label, _) in enumerate(zip(labels, values)):
        body = parts["bodies"][i]
        body.set_facecolor(COLOR_MAP.get(label, "gray"))
        body.set_alpha(0.75)
        body.set_edgecolor("black")
        body.set_linewidth(1.0)

    for i, v in enumerate(values):
        sns.stripplot(x=[i] * len(v), y=v, color="black", size=3, alpha=0.65, jitter=0.18)

    plt.xlabel("Method", fontsize=11, fontweight="bold")
    plt.ylabel("Quantile of edges per component", fontsize=11, fontweight="bold")
    plt.xticks(range(len(labels)), labels, rotation=15, fontsize=10, fontweight="bold")
    plt.yticks(fontsize=10, fontweight="bold")
    plt.ylim(0, 1.05)
    plt.grid(True, alpha=0.35)
    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"[SAVED] {save_path}")


base = os.path.join("SDAN_Comparison", "Su_2020", "output_comparison")
fig_dir = os.path.join("SDAN_Comparison", "Su_2020", "figures")
os.makedirs(fig_dir, exist_ok=True)

paths = {
    "Spectra": {
        "load": os.path.join(base, "Spectra", "train_s_cd4_cd8_BL_Spectra_union.npy"),
        "genes": os.path.join(base, "Spectra", "gene_list_cd4_cd8_union.txt"),
        "type": "spectra",
    },
    "sciRED CD4": {
        "load": os.path.join(base, "sciRED", "train_s_cd4_BL_sciRED.npy"),
        "genes": os.path.join(base, "Spectra", "gene_list_cd4_BL.txt"),
        "type": "scired",
    },
    "sciRED CD8": {
        "load": os.path.join(base, "sciRED", "train_s_cd8_BL_sciRED.npy"),
        "genes": os.path.join(base, "Spectra", "gene_list_cd8_BL.txt"),
        "type": "scired",
    },
}

results = {}
for name, cfg in paths.items():
    print(f"\n[INFO] Processing {name}")
    train_s = np.load(cfg["load"], allow_pickle=True)
    if train_s.dtype == object:
        train_s = np.array([np.asarray(row, dtype=float).ravel() for row in train_s])
    with open(cfg["genes"], "r") as f:
        genes = pd.Index([line.strip().split()[0] for line in f if line.strip()])
    if cfg["type"] == "spectra":
        train_s = normalize_rows_sum(train_s)
    edge_list_index = construct_gene_graph(genes).detach().numpy()
    size, degree, quantile = compute_connectivity_uniform_fast(train_s, genes, edge_list_index)
    results[name] = {"size": size, "degree": degree, "quantile": quantile}

quantile_dict = {name: results[name]["quantile"] for name in results}
out_path = os.path.join(fig_dir, "quantile_edge_Su2020_output_comparison.pdf")
plot_quantile_edge_violin(quantile_dict, save_path=out_path, title="BioGRID Connectivity")

```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

BASE = os.path.join("SDAN_Comparison", "Su_2020", "output_comparison")
FIGDIR = os.path.join("SDAN_Comparison", "Su_2020", "figures")
os.makedirs(FIGDIR, exist_ok=True)

paths = {
    "CD4 (sciRED)": os.path.join(BASE, "sciRED", "results_classifiers_cd4_BL_sciRED_logreg_individual_scores.csv"),
    "CD8 (sciRED)": os.path.join(BASE, "sciRED", "results_classifiers_cd8_BL_sciRED_logreg_individual_scores.csv"),
    "CD4 (SDAN)": os.path.join(BASE, "SDAN", "results_classifiers_cd4_BL_SDAN_2.0_logreg_individual_scores.csv"),
    "CD8 (SDAN)": os.path.join(BASE, "SDAN", "results_classifiers_cd8_BL_SDAN_2.0_logreg_individual_scores.csv"),
    "CD4 (Spectra)": os.path.join(BASE, "Spectra", "results_classifiers_cd4_BL_Spectra_union_Spectra_union_logreg_ind_scores.csv"),
    "CD8 (Spectra)": os.path.join(BASE, "Spectra", "results_classifiers_cd8_BL_Spectra_union_Spectra_union_logreg_ind_scores.csv"),
}

palette = sns.color_palette("coolwarm", 2)
COLOR_MAP = {"Mild": palette[0], "Severe": palette[1]}
GROUP_ORDER = ["Mild", "Severe"]


def draw_boxplot(ax, df):
    sns.boxplot(data=df, x="label_str", y="score_pos", order=GROUP_ORDER, palette=COLOR_MAP, width=0.28, linewidth=1.2, showfliers=False, boxprops=dict(alpha=0.75), ax=ax)
    sns.stripplot(data=df, x="label_str", y="score_pos", order=GROUP_ORDER, palette=COLOR_MAP, jitter=0.15, size=3, alpha=0.75, linewidth=0.3, edgecolor="black", ax=ax)
    ax.set_xlabel("Clinical Group", fontsize=11, fontweight="bold")
    ax.set_ylabel("Predicted Severity Score", fontsize=11, fontweight="bold")
    ax.set_ylim(0, 1)
    ax.tick_params(labelsize=10)


def plot_individual(df, tag):
    fig, ax = plt.subplots(figsize=(3, 3), dpi=300)
    draw_boxplot(ax, df)
    ax.set_title(tag, fontsize=12, fontweight="bold")
    out = os.path.join(FIGDIR, f"Su2020_boxplot_{tag.replace(' ', '_')}.pdf")
    plt.tight_layout()
    plt.savefig(out, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close(fig)
    print(f"[SAVED] {out}")


def plot_combined(dfs_dict):
    order = ["CD4 (sciRED)", "CD4 (SDAN)", "CD4 (Spectra)", "CD8 (sciRED)", "CD8 (SDAN)", "CD8 (Spectra)"]
    fig, axes = plt.subplots(2, 3, figsize=(7.2, 5.4), dpi=300)
    axes = axes.flatten()
    for ax, tag in zip(axes, order):
        draw_boxplot(ax, dfs_dict[tag])
        ax.set_title(tag, fontsize=10, fontweight="bold")
        ax.set_xlabel("")
        ax.set_ylabel("")
        ax.tick_params(labelsize=9)
        ax.set_ylim(0, 1)
    fig.supxlabel("Clinical Group", fontsize=13, fontweight="bold", y=0.05)
    fig.supylabel("Predicted Severity Score", fontsize=13, fontweight="bold", x=0.02)
    fig.suptitle("Predicted Severity Scores Across Methods — Su 2020", fontsize=15, fontweight="bold", y=0.98)
    plt.tight_layout(rect=[0.06, 0.06, 1, 0.93])
    out = os.path.join(FIGDIR, "Su2020_boxplot_combined.pdf")
    plt.savefig(out, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close(fig)
    print(f"[SAVED] {out}")


dfs_dict = {}
for tag, path in paths.items():
    print(f"[LOAD] {tag}: {path}")
    df = pd.read_csv(path)
    df["label_str"] = df["label_str"].str.capitalize()
    dfs_dict[tag] = df
    plot_individual(df, tag)

plot_combined(dfs_dict)

```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

BASE = os.path.join("SDAN_Comparison", "Yost_2019", "output_comparison")
OUT_FIGS = os.path.join("SDAN_Comparison", "Yost_2019", "figures")
os.makedirs(OUT_FIGS, exist_ok=True)

paths = {
    "CD8T (sciRED)": os.path.join(BASE, "sciRED", "results_classifiers_CD8T_sciRED_logreg_individual_scores.csv"),
    "CD8T (SDAN)": os.path.join(BASE, "SDAN", "results_classifiers_CD8T_SDAN_2.0_logreg_individual_scores.csv"),
    "CD8T (Spectra)": os.path.join(BASE, "Spectra", "results_classifiers_CD8T_Spectra_Spectra_logreg_ind_scores.csv"),
}

palette = sns.color_palette("coolwarm", 2)
COLOR_MAP = {"No": palette[0], "Yes": palette[1]}
GROUP_ORDER = ["No", "Yes"]


def plot_boxplot(df, title):
    fig, ax = plt.subplots(figsize=(3, 3), dpi=300)
    sns.boxplot(data=df, x="label_str", y="score_pos", order=GROUP_ORDER, palette=COLOR_MAP, width=0.25, linewidth=1.2, showfliers=False, boxprops=dict(alpha=0.75), ax=ax)
    sns.stripplot(data=df, x="label_str", y="score_pos", order=GROUP_ORDER, palette=COLOR_MAP, size=3, jitter=0.15, alpha=0.75, linewidth=0.3, edgecolor="black", ax=ax)
    ax.set_title(title, fontsize=11, fontweight="bold")
    ax.set_xlabel("Clinical Group", fontsize=11, fontweight="bold")
    ax.set_ylabel("Predicted Probability", fontsize=11, fontweight="bold")
    ax.set_ylim(0, 1)
    ax.tick_params(labelsize=10)
    plt.tight_layout()
    out = os.path.join(OUT_FIGS, f"Yost_boxplot_{title.replace(' ', '_')}.pdf")
    plt.savefig(out, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close(fig)
    print(f"[SAVED] {out}")


def plot_all_combined(dfs_dict):
    ordered_tags = ["CD8T (sciRED)", "CD8T (SDAN)", "CD8T (Spectra)"]
    fig, axes = plt.subplots(1, 3, figsize=(9, 3.4), dpi=300)
    for ax, tag in zip(axes, ordered_tags):
        df = dfs_dict[tag]
        sns.boxplot(data=df, x="label_str", y="score_pos", order=GROUP_ORDER, palette=COLOR_MAP, width=0.25, linewidth=1.2, showfliers=False, boxprops=dict(alpha=0.75), ax=ax)
        sns.stripplot(data=df, x="label_str", y="score_pos", order=GROUP_ORDER, palette=COLOR_MAP, size=3, jitter=0.15, alpha=0.75, linewidth=0.3, edgecolor="black", ax=ax)
        ax.set_title(tag, fontsize=10, fontweight="bold")
        ax.set_xlabel("")
        ax.set_ylabel("")
        ax.set_ylim(0, 1)
        ax.tick_params(labelsize=9)
    fig.supxlabel("Clinical Group", fontsize=13, fontweight="bold", y=0.08)
    fig.supylabel("Predicted Probability", fontsize=13, fontweight="bold", x=0.04)
    fig.suptitle("CD8T Response Probability Across Methods — Yost 2019", fontsize=15, fontweight="bold", y=0.98)
    plt.tight_layout(rect=[0.05, 0.12, 1, 0.92])
    out = os.path.join(OUT_FIGS, "Yost_combined_boxplot.pdf")
    plt.savefig(out, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close(fig)
    print(f"[SAVED] {out}")


dfs_dict = {}
for tag, path in paths.items():
    print(f"[LOAD] {tag}: {path}")
    df = pd.read_csv(path)
    df["label_str"] = df["label_str"].replace({"R": "Yes", "NR": "No"})
    dfs_dict[tag] = df
    plot_boxplot(df, tag)

plot_all_combined(dfs_dict)

```
